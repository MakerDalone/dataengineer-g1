```python
from datetime import datetime
import time

total_intentos = 3
n_intento = 0

errores= []

while n_intento < total_intentos:
    try:
        path = "../data/ventas_20260926.txt"
        with open(path, "r") as f:
            contenido = f.read()
            print(contenido)
    except FileNotFoundError as e:
        errores.append(f"{datetime.now()} | Error: {e}") 
    n_intento += 1   
    time.sleep(10)

print(errores)      
```
