TIPOS DE VARIABLES

```python
cantidad: int = 3
precio: float = 145.0
cliente: str = "Marco Torres"
activo: bool = True
Descuento: NoneType = None 
```

```python
print (type(cantidad))
print(type(precio))
print(type(cliente))
print(type(activo))
print(type(Descuento))  
```

CASTEO DE VARIABLES

```python
cantidad = "3"
print(type(cantidad))

cantidad = int(cantidad)
print(type(cantidad))
```

```python
precio = "130.5"
print(type(precio))

precio = float(precio)
print(type(precio))
print(precio)

precio = int(precio)
print(precio)
```

```python
activo = 0
print(activo)

activo = bool(activo)
print(activo)
```
