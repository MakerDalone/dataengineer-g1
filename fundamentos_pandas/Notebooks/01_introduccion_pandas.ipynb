SERIES Y DATAFRAME

```python
#Una SERIE Es un array ID
serie = pd.Series(
    [10,20,30],
    index=["a", "b", "c"],
    name="ventas"
)
serie

#Un DATAFRAME es un array 2D
df= pd.DataFrame({
    "producto" : ["A", "B"],
    "unidades" : [10, 20]
})
```

```python
serie_ventas = pd.Series(
    [120,85,200,45],
    index=["enero", "febrero", "marzo", "abril"],
    name = "ventas"
)
serie_ventas
```
